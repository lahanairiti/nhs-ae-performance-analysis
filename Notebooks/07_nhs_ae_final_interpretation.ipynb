# NHS A&E Performance Analysis – Final Report

## Project Objective

The aim of this project was to analyse NHS England's monthly A&E statistics to understand attendance volumes and four-hour performance across Type 1 A&E departments.

The analysis covers April 2025 to March 2026.

Python and SQL were used for data cleaning, analysis and KPI calculations, while Power BI was used to visualise the results.


## Key Performance Indicators

The project focused on the following measures:

1. Total A&E attendances
2. Attendances exceeding four hours
3. Attendances within four hours
4. Percentage of attendances within four hours
5. Monthly attendance trends
6. Provider-level performance comparisons

## Key Findings

### Monthly Performance

- Four-hour performance varied throughout the reporting year.
- Performance increased from 59.79% in April 2025 to 62.70% in July 2025.
- It subsequently declined, reaching its lowest point of 57.04% in January 2026.
- Performance recovered in February and March, reaching 63.89% in March.
- March recorded the highest four-hour performance, while January recorded the lowest.

### Provider Comparison

- University Hospitals Birmingham NHS Foundation Trust had the highest annual attendance volume among the providers displayed.
- Royal Free London NHS Foundation Trust recorded the highest four-hour performance among the providers displayed, at 66.43%.
- Barts Health NHS Trust recorded the lowest performance among those displayed, at 55.95%.
- The difference between these two provider percentages was 10.48 percentage points.

## Limitations

1. The analysis focuses on Type 1 A&E departments.
2. The data is aggregated monthly and does not contain individual patient information.
3. Provider comparisons are not adjusted for patient characteristics, staffing or hospital capacity.
4. Missing data may affect comparisons between months and providers.
5. The analysis covers only one financial year.
6. The findings identify trends and associations, not the causes of waiting-time delays.

## Recommendations

1. Monitor attendance volumes alongside four-hour performance.
2. Investigate providers showing sustained differences in performance.
3. Monitor missing data and reporting completeness.
4. Extend the analysis using other publicly available NHS operational indicators to investigate potential factors associated with waiting-time delays.

## Conclusion

This project provided an opportunity to work with real-world NHS healthcare statistics and apply data analysis techniques to an operational healthcare topic.

Using Python and SQL, I explored and validated the data, calculated key performance indicators and examined monthly and provider-level trends.

Power BI was used to present the findings in an interactive dashboard.

The project strengthened my understanding of healthcare data quality, KPI reporting, data visualisation and communicating analytical findings.
